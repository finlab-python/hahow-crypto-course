## 建立 DataFrame

[DataFrame 文件](https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.html)

- **data**：資料，可以是 dict、list、numpy array 或另一個 DataFrame。
- **index**：每一列（row）的名稱，預設是 0, 1, 2...

```python
import pandas as pd

strategy = pd.DataFrame({'annual_return': [0.2, 0.4, 0.5], 'risk': [0.3, 0.4, 0.5]}, index=['sma', 'rsi', 'kd'])
strategy
```

## Series

取出 DataFrame 的一個欄位，就是一個 Series。

```python
strategy['risk']
```

## 運算

欄位之間可以直接運算，例如報酬 ÷ 風險（類似夏普率）：

```python
strategy['annual_return'] / strategy['risk']
```

## 邏輯判斷

```python
# 風險 <= 0.4 嗎？
strategy.risk <= 0.4
```

## 篩選列

```python
# 只留下風險 <= 0.4 的策略
strategy[strategy.risk <= 0.4]
```

## 改變資料形狀

[melt 文件](https://pandas.pydata.org/docs/reference/api/pandas.melt.html)

```python
strategy.melt()
```
